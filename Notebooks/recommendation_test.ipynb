```python
import joblib

knn = joblib.load('C:/Users/hi/Documents/PROJEK BELAJAR/machine learning/Nutri_Match/models/knn_model.pkl')
scaler = joblib.load('C:/Users/hi/Documents/PROJEK BELAJAR/machine learning/Nutri_Match/models/scaler.pkl')
```

```python
import pandas as pd

df = pd.read_csv("../data/raw/nutrition.csv")
df

print(df.shape)
print(type(scaler))
print(type(knn))
```

```python
import numpy as np

input_nutrition = np.array([
    [300, 15, 10, 40]
])

input_scaled = scaler.transform(input_nutrition)

distances, indices = knn.kneighbors(input_scaled)
print(indices)
print(distances)
```

```python
recommendations = df.iloc[indices[0]].copy()

recommendations['distances'] = distances[0]

recommendations

df.columns

result = recommendations[
    [
        'id',
        'calories',
        'proteins',
        'fat',
        'carbohydrate',
        'name',
        'image',
        'distances'
    ]
]

result

```

```python
def recommend_food(calories, proteins, fat, carbohydrate, n=5):
    
    input_nutrition = np.array([
        [calories, proteins, fat, carbohydrate]
    ])
    
    input_scaled = scaler.transform(input_nutrition)
    
    distances, indices = knn.kneighbors(
        input_scaled,
        n_neighbors=n
    )
    
    recommendations = df.iloc[indices[0]].copy()
    
    recommendations['distances'] = distances[0]
    
    return recommendations

recommend_food(300, 15, 10, 40)
recommend_food(200, 10, 5, 25)
recommend_food(500, 25, 20, 60)
```

```python
test_inputs = [
    [200, 10, 5, 25],
    [300, 15, 10, 40],
    [500, 25, 20, 60],
    [700, 30, 30, 80]
]

for i, nutrition in enumerate(test_inputs, 1):
    print(f"\n===== TEST {i} =====")
    print(
        f"Input: {nutrition[0]} kcal, "
        f"{nutrition[1]} protein, "
        f"{nutrition[2]} fat, "
        f"{nutrition[3]} carbohydrate"
    )

    result = recommend_food(*nutrition)

    print(
        result[
            ["name", "calories", "proteins",
             "fat", "carbohydrate", "distances"]
        ].to_string(index=False)
    )
    
    
```
