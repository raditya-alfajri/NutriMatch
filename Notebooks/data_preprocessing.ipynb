```python
import pandas as pd
import numpy as np

data = pd.read_csv(r'C:\Users\hi\Documents\PROJEK BELAJAR\machine learning\Nutri_Match\data\raw\nutrition_categorized_v2.csv')

features = ["calories", "proteins", "fat", "carbohydrate"]

# Ambil fitur nutrisi
x = data[features]

# Hitung IQR
Q1 = x.quantile(0.25)
Q3 = x.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Deteksi outlier
outlier = (x < lower_bound) | (x > upper_bound)

outlier
data
```
