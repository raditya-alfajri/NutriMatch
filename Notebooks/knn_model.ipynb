```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler 

data = pd.read_csv(r'C:\Users\hi\Documents\PROJEK BELAJAR\machine learning\Nutri_Match\data\raw\nutrition_categorized_v2.csv')

features = ["calories", "proteins", "fat", "carbohydrate"]

# Ambil fitur nutrisi
X = data[features].dropna()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import joblib

joblib.dump(
    scaler,
    "../models/scaler.pkl"
)
```

```python
from sklearn.neighbors import NearestNeighbors
import joblib

knn = NearestNeighbors(
    n_neighbors=10,
    metric="euclidean"
)

knn.fit(X_scaled)
joblib.dump(knn, 'C:/Users/hi/Documents/PROJEK BELAJAR/machine learning/Nutri_Match/models/knn_model.pkl')
```

```python
print(data[features].isnull().sum())
```

```python
input_nutrition = np.array([
    [300, 15, 10, 40]
])

input_scaled = scaler.transform(input_nutrition)

distances, indices = knn.kneighbors(input_scaled)
print(indices)
print(distances)
```

```python
recommendations = data.iloc[indices[0]].copy()

recommendations["distance"] = distances[0]

result = recommendations[
    [
        'id',
        'name',
        'calories',
        'proteins',
        'fat',
        'carbohydrate',
        'category_v2',
        'distance'
    ]
]

result
```
