```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler 

data = pd.read_csv(r'C:\Users\hi\Documents\PROJEK BELAJAR\machine learning\Nutri_Match\data\raw\nutrition.csv')

features = ["calories", "proteins", "fat", "carbohydrate"]

# Ambil fitur nutrisi
X = data[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.neighbors import NearestNeighbors
import joblib

knn = NearestNeighbors(
    n_neighbors=5,
    metric="euclidean"
)

knn.fit(X_scaled)
joblib.dump(knn, 'C:/Users/hi/Documents/PROJEK BELAJAR/machine learning/Nutri_Match/models/knn_model_pk1.ipynb')
joblib.dump(scaler, 'C:/Users/hi/Documents/PROJEK BELAJAR/machine learning/Nutri_Match/models/scaler.pkl')
```
